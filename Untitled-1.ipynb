### Database analysis

```python
import sqlite3   
import pandas as pd
```

```python
conn = sqlite3.connect('northwind.db')
tables = pd.read_sql("SELECT name FROM sqlite_master WHERE type='table';", conn)
print(tables)   
```

```python
views = pd.read_sql("SELECT name FROM sqlite_master WHERE type='view';", conn)
print(views)  
```

```python
# Print contents of every table
for table in tables['name']:
    print(f"\n{'='*50}")
    print(f"TABLE: {table}")
    print('='*50)
    df = pd.read_sql(f'SELECT * FROM "{table}"', conn)
    print(f"Shape: {df.shape[0]} rows x {df.shape[1]} cols")
    print(df.head())

```

```python
for table in tables['name'].tolist():
    fk = pd.read_sql(f'PRAGMA foreign_key_list("{table}");', conn)
    if not fk.empty:
        print(f"\n{table} → connects to:")
        print(fk[['table', 'from', 'to']].to_string(index=False))
```

### Q1) Which product category generates the most revenue, sort the categories in descending order on the basis of revenue.

```python
categoryWiseRevenue = pd.read_sql("""-- Layer 1: Just get the raw data first
SELECT c.CategoryName, od.UnitPrice, od.Quantity, od.Discount
FROM "Order Details" od
JOIN Products p ON od.ProductID = p.ProductID
JOIN Categories c ON p.CategoryID = c.CategoryID;""",conn)

print(categoryWiseRevenue)
```

```python
categoryWiseRevenue = pd.read_sql(""" -- Layer 2: Add the calculation
SELECT c.CategoryName,
       od.UnitPrice * od.Quantity * (1 - od.Discount) AS line_revenue
FROM "Order Details" od
JOIN Products p ON od.ProductID = p.ProductID
JOIN Categories c ON p.CategoryID = c.CategoryID;""",conn                               
                                  )

print(categoryWiseRevenue)
```

```python
categoryWiseRevenue = pd.read_sql("""
-- Layer 3: Aggregate and sort
SELECT c.CategoryName,
       ROUND(SUM(od.UnitPrice * od.Quantity * (1 - od.Discount)), 2) AS revenue
FROM "Order Details" od
JOIN Products p ON od.ProductID = p.ProductID
JOIN Categories c ON p.CategoryID = c.CategoryID
GROUP BY c.CategoryName
ORDER BY revenue DESC;""",conn)

print(categoryWiseRevenue)
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=categoryWiseRevenue, x='revenue', y='CategoryName', palette='viridis')

def smart_formatter(x, pos):
    if x >= 1e9:
        return f'${x/1e9:.1f}B'
    elif x >= 1e6:
        return f'${x/1e6:.0f}M'
    elif x >= 1e3:
        return f'${x/1e3:.0f}K'
    else:
        return f'${x:.0f}'

plt.gca().xaxis.set_major_formatter(mticker.FuncFormatter(smart_formatter))

plt.title('Revenue by Product Category', fontsize=14)
plt.xlabel('Revenue ($)')
plt.ylabel('Category')
plt.tight_layout()
plt.show()
```

### Analysis:-

1) Beverages is the clear revenue leader, generating roughly $9.2M - significantly ahead of the second-place category, Confections (~$6.6M).
2) The top 3 categories (Beverages, Confections, Meat/Poultry) together account for a disproportionate share of total revenue compared to the bottom 3.
3) The gap between the highest and lowest performing category is roughly 3.2x, indicating meaningful variation in category performance rather than uniform demand.

### Recommendations:-

1) Prioritize inventory and marketing investment in Beverages and Confections. These categories are proven demand drivers and likely have the highest ROI on promotional spend.
2) Investigate why Grains/Cereals underperforms. This could stem from pricing, limited SKU variety, low customer awareness, or genuine low demand. A follow-up analysis on order frequency vs. average order value for this category would clarify the cause.
3) Consider cross-selling/bundling strategies. Pairing high-performing categories (Beverages) with low-performing ones (Grains/Cereals) in promotional bundles could lift weaker categories without heavy standalone marketing spend.


### Q2) How does quarterly revenue trend over time?

```python
quarterlyRevenue = pd.read_sql("""
SELECT 
STRFTIME('%Y', o.OrderDate) AS Year,
    CASE 
        WHEN STRFTIME('%m', o.OrderDate) IN ('01','02','03') THEN 'Q1'
        WHEN STRFTIME('%m', o.OrderDate) IN ('04','05','06') THEN 'Q2'
        WHEN STRFTIME('%m', o.OrderDate) IN ('07','08','09') THEN 'Q3'
        ELSE 'Q4'
    END AS Quarter, 
ROUND(SUM(od.UnitPrice * od.Quantity * (1 - od.Discount)), 2) AS Revenue
FROM "Order Details" od
JOIN Orders o ON od.OrderID = o.OrderID
GROUP BY Year, Quarter
""",conn)

print(quarterlyRevenue)
```

```python
quarterlyRevenue['Period'] = quarterlyRevenue['Year'] + '-' + quarterlyRevenue['Quarter']

plt.figure(figsize=(15, 8))
sns.barplot(data=quarterlyRevenue, x='Period', y='Revenue', palette='viridis')

def smart_formatter(y, pos):
    if y >= 1e9:
        return f'${y/1e9:.1f}B'
    elif y >= 1e6:
        return f'${y/1e6:.0f}M'
    elif y >= 1e3:
        return f'${y/1e3:.0f}K'
    else:
        return f'${y:.0f}'

plt.gca().yaxis.set_major_formatter(mticker.FuncFormatter(smart_formatter))

plt.title('Quarterly Revenue', fontsize=14)
plt.xlabel('Quarter')
plt.ylabel('Revenue ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
